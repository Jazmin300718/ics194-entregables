**1.Base de datos**

La base de datos seleccionada corresponde a "hprice2" proveniente del paquete wooldridge, la cual contiene informacion de factores determinantes de los precios de las viviendas. La base de datos cuenta con un total de 506 observaciones las que corresponden al numero de casas o barrios, y un total de 12 variables entre las cuales las que mas se destacan corresponden a numero de habitaciones, tasas de criminalidad, nivel de contaminacion y porcentaje de poblacion de bajo estatuts socieconomico.






```r
install.packages("wooldridge")
library(wooldridge)
data("hprice2")
head(hprice2)
names(hprice2)
str(hprice2)
```

**2. Análisis Descriptivo**

En la siguiente celda se pueden observar estadisticas descriptivas de las principales variables del modelo.



```r
summary(hprice2)
```

A partir de las estadísticas descriptivas, observamos que el precio promedio de las viviendas es 22.512, con una mediana de 21.200, lo que nos indica una distribución equilibrada, aunque con presencia de valores extremos.
Sobre la dispersión, el precio presenta un rango amplio (desde 5.000 hasta 50.001), lo que indica una alta variabilidad entre las viviendas analizadas. Por lo que concluimos que existen diferencias importantes en las características de las propiedades.

La variable criminalidad (crime) presenta una media (3.61) mayor que la mediana (0.256), lo que evidencia una asimetría positiva. Esto nos indica la presencia de valores extremos altos que influyen en el promedio.

Por otro lado, variables como el número de habitaciones (rooms) muestran menor dispersión, ya que sus valores se concentran en un rango más acotado (entre 3.5 y 8.8), lo que sugiere mayor similitud en el tamaño de las viviendas.

**3. Análisis Gráfico**

A continuacion analizaremos algunos graficos de las variables mas relevantes de este modelo.  



```r
hist(hprice2$price, col="pink", main="Distribución del precio")
```

Observando el histograma de los precios se puede concluir que el mercado inmobiliario es bastante diverso. Esas casas, la mayoría, están concentradas en rangos de precio moderados, pero hay un pequeño grupo de propiedades que tienen valores excepcionalmente altos y eso inclina la balanza hacia la derecha, esa “cola” o asimetría positiva en los datos. Esa gran dispersión confirma que no nos encontramos frente a un mercado uniforme, sino muy heterogéneo. Estos valores extremos pueden distorsionar los resultados, por lo que lo ideal es aplicar una transformación logarítmica al precio, con lo que se estabilizan los datos y se hace que el modelo econométrico sea mucho más preciso y confiable.

```r
plot(hprice2$rooms, hprice2$price, col="green",
     main="Relacion Precio|Habitaciones",
     xlab="Habitaciones", ylab="Precio")
```

Al observar el gráfico de dispersión, podemos ver claramente que existe una relación directa entre el precio y la cantidad de habitaciones: a mayor número de habitaciones, el precio tiende a aumentar, lo cual tiene sentido ya que la cantidad de habitaciones suele ser un buen indicador del tamaño y la calidad de la vivienda. La tendencia general está bastante clara, pero los puntos están algo dispersos, recordándonos que el valor de una propiedad depende de muchos otros factores aparte de éste. En cualquier caso, esta relación confirma que el número de habitaciones es un factor determinante, por lo que es una variable fundamental que debemos incluir en nuestro modelo para explicar mejor la fijación de precios.

```r
plot(hprice2$crime, hprice2$price, col="blue",
     main="Relacion Precio|Criminalidad",
     xlab="Criminalidad", ylab="Precio")
```

En el gráfico que relaciona el precio de las viviendas con la tasa de criminalidad, se puede ver que existe una relación negativa entre estas dos variables. O sea, a mayor nivel de criminalidad en una zona, menor precio tiende a tener la vivienda. Ese comportamiento se explica porque las zonas donde existen mayores niveles de inseguridad son las menos atractivas para los compradores, lo que reduce la demanda y, por lo tanto, el valor de los inmuebles. Pero el gráfico también muestra una gran dispersión, sobre todo en los niveles bajos de criminalidad, lo que indica que este factor, aunque importante, no es el único determinante del precio.

```r
plot(hprice2$nox, hprice2$price,
     main="Precio vs Contaminación (NOX)",
     xlab="Nivel de contaminación (nox)",
     ylab="Precio de la vivienda",
     col="darkgreen", pch=16)

abline(lm(price ~ nox, data=hprice2), col="orange", lwd=2)
```

Se observa una relación negativa entre el nivel de contaminación y el precio de las viviendas. A medida que aumenta la contaminación del aire (medida por NOX), el precio tiende a disminuir, lo cual se ve reforzado por la línea de tendencia descendente incorporada en el gráfico.

Esto es consistente con la teoría económica, ya que las personas valoran vivir en entornos con mejor calidad ambiental, lo que reduce la demanda por viviendas en esas zonas. Esta relación parece más estable que otras variables, lo que sugiere que el nivel de contaminación es un determinante importante en la valoración de las propiedades, lo que se traduce en precios más altos en zonas menos contaminadas.


```r
plot(hprice2$lowstat, hprice2$price,
     main="Precio vs Nivel socioeconómico (lowstat)",
     xlab="% población de bajo estatus",
     ylab="Precio de la vivienda",
     col="purple", pch=16)

abline(lm(price ~ lowstat, data=hprice2), col="blue", lwd=2)
```

El gráfico exhibe la correlación existente entre el precio de la vivienda y el porcentaje de la población de bajo estatus socioeconómico (lowstat). Se puede apreciar una tendencia inversa, puesto que a medida que incrementa el porcentaje de población menos favorecida de nivel socioeconómico, el precio de las viviendas disminuye, lo cual se puede ver reflejado en el hecho de que la pendiente de la recta de regresión es negativa.
Esto se da por el hecho de que los sectores más vulnerables tienden a tener baja demanda, escaso acceso a servicios y una escasa atracción, lo que repercute en el precio de las viviendas. Aun así, la dispersión de los datos nos muestra que los precios de las viviendas no solo dependen de esta variable, sino que también influyen otros factores como por ejemplo la criminalidad, la contaminación, las características de las viviendas, etc.
